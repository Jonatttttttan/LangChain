Sequential Chains

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
```

```python
llm1 = ChatOpenAI(temperature=0.5, tiktoken_model_name="gpt-3.5-turbo")
```

```python
prompt_template1 = PromptTemplate.from_template(
    template = "Eres un científico experimentado y programador en Python. Escribe una función con la sintaxe de  Python que implemente el concepto de {concepto}"
)
```

```python
chain1 = prompt_template1 | llm1
```

```python
llm2  = ChatOpenAI(tiktoken_model_name="gpt-3.5-turbo", temperature=1.2)
```

```python
prompt_template2 = PromptTemplate.from_template(
    template = "Dada la función de Python {input}, descríbela matematicamente."
)
```

```python
chain2 = prompt_template2 | llm2
```

```python
chain_sequential = prompt_template1 | llm1 | prompt_template2 | llm2
```

```python
output = chain_sequential.invoke( {'concepto' : 'obtener los primeros "n" números primos'})
print(output)
```

ReAct Agent (RAG)
Imagina que tienes un robot virtual que puede hacer tareas por ti en internet, como buscar la última notícia, revisar el clima, o incluso encontrar el perfil de Linkedin de alguien. Ese robot es lo que llamamos un "agente" en LangChain. Pero este robot es especial, porque no solo sigue ordenes simples; primero piensa en como resolver tu petición, decide los pasos a seguir, y luego los ejecuta, como buscar en Google o consultar una base de dados. es como darle superpoderes a los modelos de lenguaje, esos programas que entienden y generan texto, permitiéndoles interactuar con el mundo real u traerte la información que necesitas. Y lo más interesante es que puedes enseñarle a tu robot a hacer cosas nuevas, dándole acceso a diferentes herramientas u servicios en internet

Crear a ReAct Agent

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.tools import Tool
from langchain_community.tools import DuckDuckGoSearchRun, WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
from langchain_experimental.tools.python.tool import PythonREPLTool
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0)
```

```python
template = 'Responde las siguientes preguntas en Italiano lo mejor que puedas. Preguntas: {q}'
```

```python
prompt_template = PromptTemplate.from_template(template)
```

```python
prompt = "O que é fibonacci?"
```

```python
#1. Python REPL Tool
python_repl = PythonREPLTool()
python_repl_tool = Tool(
    name = 'Python REPL',
    func=python_repl.run,
    description='Útil cuando necesitas usar Python para responder una pergunta. Debes ingresar código Python.'
)
```

```python
# 2. Wikipedia Tool
api_wrapper = WikipediaAPIWrapper()
wikipedia = WikipediaQueryRun(api_wrapper=api_wrapper)
wikipedia_tool = Tool(
    name = "Wikipedia",
    func=wikipedia.run,
    description='Útil cuando necesitas buscar un tema, país o persona en Wikipedia.'
)
```

```python
# 3. DuckDuckGo Search Tool
search = DuckDuckGoSearchRun()
duckduckgo_tool = Tool(
    name = 'DuckDuckGo Search',
    func=search.run,
    description='Útil cuando necesitas realizar una búsqueda en internet para encontrar información que otra herramienta que otra no puede proporcionar.'
)
```

```python
system_prompt = (
    "Você é um agente útil. Responda SEMPRE em italiano.\n"
    "Quando for útil, use a tool Wikipedia para fatos e Python REPL para cálculos.\n"
    "Se usar uma ferramenta, incorpore o resultado na resposta final em italiano."
)
```

```python
tools = [PythonREPLTool]
```

```python
agent = create_agent(model=llm, tools=tools, system_prompt=system_prompt)
```
